```python
# find outliers: points outside the lower boundary is -ve outlier and upper is +ve
# lower boundary= Q1 - (1.5 * IQR)
# upper boundary=  Q3 + (1.5 * IQR)
```

```python
lower_boundary= Q1 - (1.5 * IQR)
upper_boundary=  Q3 + (1.5 * IQR)
```

```python
import matplotlib.pyplot as plt
```

```python
sns.boxplot(data= df, x= "total_bill")
plt.show()
```

```python
# +ve outliers present
# if normally distributed ddata is present and it has some outliers need to handle them using IQR
```

```python
Q1 = df['total_bill'].quantile(0.25)
Q3 = df['total_bill'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df = df[
    (df['total_bill'] >= lower_limit) &
    (df['total_bill'] <= upper_limit)
]
```

```python
sns.boxplot(data= df, x= "total_bill")
plt.show()
```
