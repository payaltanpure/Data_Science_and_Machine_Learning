```python
# 1. First understand (353, 1)

# A shape is written as:

# (number of rows, number of columns)

# So:

# (353, 1)

# means:

# 353 rows/data records
# 1 column/feature
# 2. Your four outputs
# X_train.shape
# (353, 1)

# Means:

# 353 training records and 1 input feature.

# For example:

# X_train

#    X
#    ↓
# [ 10 ]
# [ 20 ]
# [ 15 ]
# [ 30 ]
# ...
# [353 rows]
# X_test.shape
# (89, 1)

# Means:

# 89 testing records and 1 input feature.

# X_test

#    X
#    ↓
# [ 25 ]
# [ 18 ]
# [ 40 ]
# ...
# [89 rows]
# Y_train.shape
# (353, 1)

# Means:

# 353 actual target values corresponding to the 353 training inputs.

# X_train              Y_train

# 10  ───────────────→  50
# 20  ───────────────→  70
# 15  ───────────────→  60
# 30  ───────────────→  90

# So:

# $$ \boxed{X_{train} \leftrightarrow Y_{train}} $$

# They have the same 353 rows because each X must have its corresponding Y.

# Y_test.shape
# (89, 1)

# Means:

# 89 actual target values corresponding to the 89 test inputs.

# X_test              Y_test

# 25  ──────────────→  75
# 18  ──────────────→  55
# 40  ──────────────→  95
# 3. Why 353 + 89 = 442?

# Your complete dataset has:

# $$ 353+89=442 $$

# So you originally have 442 records.

# You specified:

# test_size=0.2

# meaning 20% for testing.

# 20% of 442:

# $$ 442\times0.2=88.4 $$

# Since we cannot have 0.4 of a row, it becomes 89 test rows.

# Remaining:

# $$ 442-89=353 $$

# So:

#              442 TOTAL RECORDS
#                     │
#           ┌─────────┴─────────┐
#           ↓                   ↓
#        80%                  20%
#     TRAINING               TESTING
#        353                   89
#         │                     │
#         ↓                     ↓
#    X_train + Y_train     X_test + Y_test
#     (353,1)  (353,1)      (89,1)  (89,1)
# 4. Why is there 1 after the comma?

# Because you have one feature/input column.

# For Simple Linear Regression, you have only one X variable.

# For example:

# Hours Studied
# 2
# 4
# 6
# 8

# Then:

# X.shape

# would be:

# (4, 1)

# because:

# 4 rows
# 1 feature

# If you had 3 input features, it could look like:

# (442, 3)

# For example:

# Age	Experience	Education
# 22	1	16
# 25	3	16
# 28	5	18

# Then there are:

# 442 rows
# 3 columns/features
# ⭐ Most important thing to remember

# Your output:

# ((353, 1), (89, 1), (353, 1), (89, 1))

# means:

# Variable	Meaning
# X_train	353 input values used for training
# X_test	89 input values kept for testing
# Y_train	353 actual answers used during training
# Y_test	89 actual answers kept for comparison during testing

# And the 89 test rows are unseen by the model during training.

# Later:

# model.fit(X_train, Y_train)

# ⬆️ Model learns from 353 X + 353 Y

# Then:

# Y_pred = model.predict(X_test)

# ⬆️ Model receives the 89 unseen X values and produces 89 predicted Y values.

# Finally:

# Y_test       vs       Y_pred
# Actual                Predicted
#   89 values             89 values

# That's where your actual vs predicted comparison happens. 🎯
```

```python
def adjusted_r2(r2, n, k):
    return 1 - ((1 - r2) * (n - 1) / (n - k - 1))
```

```python
# multiple linear reg: adjusted r2  is used when there are multiple x independent features 
```

```python

X= df.drop('target', axis= 1) # all features passed as X as it is multiple reg except target
Y= df['target']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split( X, Y, test_size=0.2, random_state=42)
```

```python
X_train.shape, X_test.shape, Y_train.shape, Y_test.shape
```

```python
lr_multi = LinearRegression()
```

```python
lr_multi.fit(X_train, Y_train)
```

```python
y_pred_multi_test= lr_multi.predict( X_test)
y_pred_multi_train= lr_multi.predict( X_train)
```

```python
# training data 
print("MSE:", mean_squared_error(Y_train, y_pred_multi_train))
print("MAE:", mean_absolute_error(Y_train, y_pred_multi_train))
print("RMSE:", np.sqrt(mean_squared_error(Y_train, y_pred_multi_train)))
print("R2:", r2_score(Y_train, y_pred_multi_train))

# function defined upside is called with values of R2, n and k 
print(
    "Adjusted R2:",
    adjusted_r2(
        r2_score(Y_train, y_pred_multi_train),
        X_train.shape[0],
        X_train.shape[1]
    )
)

# Here:

# n = number of observations
# k = number of independent variables
# r2 = ordinary R2

# The adjusted 
# R2 is useful for multiple regression because it penalizes adding unnecessary features.
```

```python
# testing data 
print("MSE:", mean_squared_error(Y_test, y_pred_multi_test))
print("MAE:", mean_absolute_error(Y_test, y_pred_multi_test))
print("RMSE:", np.sqrt(mean_squared_error(Y_test, y_pred_multi_test)))
print("R2:", r2_score(Y_test, y_pred_multi_test))

print(
    "Adjusted R2:",
    adjusted_r2(
        r2_score(Y_test, y_pred_multi_test),
        X_test.shape[0],
        X_test.shape[1]
    )
)
```
