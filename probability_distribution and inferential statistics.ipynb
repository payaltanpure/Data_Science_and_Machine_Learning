```python
sample_mean= 100
population_mean= 90
population_std= 10
sample_size= 60
alpha= 0.05
n=50
```

```python
import scipy.stats as stats
```

```python
zscore = sample_mean - population_mean / (population_std / np.sqrt(n))
```

```python
zscore
```

```python
zcritical = stats.norm.ppf(1-alpha)
# here we calculated the accepted region boundary that is zcritical value for the given alpha value
# normal = normal distribution is used to find the critical value for the z-test. The ppf (percent point function) is the inverse of the cumulative distribution function (CDF) and is used to find the critical value corresponding to a given significance level (alpha). In this case, we are finding the critical value for a one-tailed test with a significance level of 0.05.


# For a right-tailed test:

# zcritical is the x-coordinate separating acceptance and rejection regions.
# The area to the right of zcritical is alpha = 0.05 that is marginal error. The area to the left of zcritical is 1-alpha = 0.95 that is confidence level.
# zscore is the observed test statistic.

# if zscore lies in the rejection region (i.e., zscore > zcritical), we reject the null hypothesis. Otherwise, we fail to reject the null hypothesis.
```

```python
if zscore > zcritical:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

```python
p_value= 1- stats.norm.cdf(zscore)
p_value
```

```python
if p_value > 0.05:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

```python
# chi- square test: used to determine whether there is a significant association between two categorical variables. It is appropriate when the data is in the form of a contingency table, which shows the frequency of observations for each combination of the two variables.
```

```python
data= sns.load_dataset("tips")
data.head()
```

```python
data[['day', 'time']]
```

```python
data_table= pd.crosstab(data['day'], data['time'])
data_table
```

```python
observed_data= data_table.values
```

```python
observed_data
```

```python
stats_test, p_value, dof, expected_freq = stats.chi2_contingency(observed_data)
```

```python
stats_test, p_value, dof, expected_freq
```

```python
chi2_test=sum([((o-e)**2)/e for o,e in zip(observed_data, expected_freq)])
chi2_test
```

```python
chi2_test_f= chi2_test[0] + chi2_test[1]
chi2_test_f
```

```python
alpha = 0.05
chi2_critical = stats.chi2.ppf(1-alpha, dof)
chi2_critical
```

```python
if chi2_test_f > chi2_critical:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

```python
p_value = 1- stats.chi2.cdf(chi2_test_f, dof)
p_value
```

```python
if pvalue < 0.05:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```
