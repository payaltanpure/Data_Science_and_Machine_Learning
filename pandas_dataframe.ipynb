```python
# merge- combine two dataframes based on common column or index
# sql join- inner, left, right, outer
```

```python
df1= pd.DataFrame({'id':[1,2,3,4], 'name':['abc', 'pqr', 'xyz', 'lmn']})
df2= pd.DataFrame({'id':[1,2,3,5], 'score':[23,24,12, 45]})
df3= pd.DataFrame({'id':[1,2,3,4], 'score':[23,24,12, 45]})
df4= pd.DataFrame({'id':[1,2,3,5], 'score':[23,24,12, 45]})
```

```python
#1. merge(left table, right table, how= inner, left, right, outer, on= common column name)
# like sql there is no comp that two tables to be joined should have same column name to join 2 tables to give in on cond , if not then we can use left_on and right_on to specify the column name of left and right table
```

```python
merged_df= pd.merge(df1, df2, on='id', how='inner')
merged_df
```

```python
merged_df= pd.merge(df1, df3, on='id', how='inner')
merged_df
```

```python
merged_df= pd.merge(df2, df4, on='id', how='inner')
merged_df
```

```python
merged_df= pd.merge(df1, df2, on='id', how='left')
merged_df
```

```python
merged_df= pd.merge(df2, df3, on='id', how='right')
merged_df
```

```python
# if two tables has differnet label of common column
```

```python
df5= pd.DataFrame({'emp_id':[1,2,3,4], 'dept':['Comp', 'IT', 'Civil', 'Comp']})
```

```python
merged_df= pd.merge(df1, df5, how='inner', left_on='id', right_on='emp_id')
merged_df
```

```python

df= df1.merge(df5, how='inner', left_on='id', right_on='emp_id')
df
```

```python
# 2. join function
# joins two tables by index value
```

```python
df1= pd.DataFrame({'sid':[1,2,3,4], 'name':['abc', 'xyz', 'mnp', 'yhu']}, index=[1,2,3,4])
df2= pd.DataFrame({'tid':[1,2,3,5],'marks':[34,78,89,78]}, index=[1,2,3,9])
df3= pd.DataFrame({'nid':[1,2,3,8],'tmarks':[34,78,89,78]}, index=[1,2,3,10])
```

```python

df3= df1.join(df2, how='inner')
df3
```

```python
df3= df1.join(df2, how='left')
df3
```

```python
df3= df1.join(df2, how='right')
df3
```

```python
# join more than two df at a time - we can use only inner and outer joins and left here no  rightjoin becoiz at right there are more than 1 table
# bydefault - left join
df= df1.join([df2, df3])
df
```

```python
df= df1.join([df2, df3], how='outer')
df
```

```python
df= df1.join([df2, df3], how= 'left')
df
```

```python
df= df1.join([df2, df3], how= 'inner')
df
```
